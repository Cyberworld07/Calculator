```python
#to find a number is even or odd
num = int(input("Enter a number: "))
if num % 2 == 0:
    print(num, "is an even number")
else:
    print(num, "is an odd number")
```

```python
# to determine if the no. is positive, negative or zero
num = float(input("Enter a number: "))
if num > 0:
    print("positive")
elif num < 0:
    print("negative")
else:
    print("zero")
```
